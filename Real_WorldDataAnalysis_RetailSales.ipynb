```python
# Real-World Data Analysis: Retail Sales

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("sales.csv")

# 1. Basic information
print(df.head())
print("\nDataset Info:")
print(df.info())

# 2. Clean data
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

# 3. Statistical summary
print("\nSummary:")
print(df.describe())

# 4. Total sales
print("\nTotal Sales:", df["Sales"].sum())

# 5. Sales by category
category_sales = df.groupby("Category")["Sales"].sum()
print("\nSales by Category:\n", category_sales)

# 6. Visualization
category_sales.plot(kind="bar", title="Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()

# 7. Sales trend
df["Date"] = pd.to_datetime(df["Date"])
daily_sales = df.groupby("Date")["Sales"].sum()

daily_sales.plot(figsize=(10,5), title="Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()

# 8. Conclusion
print("\nConclusion:")
print("The analysis identifies sales trends and the highest-performing category.")
```
